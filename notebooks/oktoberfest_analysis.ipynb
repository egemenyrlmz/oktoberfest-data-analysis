```python
df_price = pd.read_csv('beer_price.csv')
print(df_price.shape)
df_price.head()
```

```python
print(df_visitors.shape)
df_visitors.tail()
```

```python
df_merged = pd.merge(df_visitors, df_price, on="Year", how="outer")
print(df_merged.shape)
df_merged
```

```python
df_merged_indicator = pd.merge(df_visitors, df_price, on="Year", how="outer", indicator=True)
df_merged_indicator.head(10)
```

```python
df_merged_indicator["_merge"].value_counts()
```

```python
RATE = 1.95583
f = df_merged["Year"] <= 2001
df_updated = df_merged.copy()
df_updated.loc[f, ["Min price", "Max price"]] = (
    df_updated.loc[f, ["Min price", "Max price"]] / RATE
)
```

```python
df_updated[df_updated["Year"].between(1995, 2005)]
```

```python
f = df_updated["Year"].between(2000, 2007)

mean = df_updated.loc[f, "Min price"].mean()
variance = df_updated.loc[f, "Min price"].var()

print("mean =", mean)
print("var =", variance)
```

```python
df_updated["Price Increase (%)"] = (
    (df_updated["Min price"] - df_updated["Min price"].shift(1))
    / df_updated["Min price"].shift(1)
    * 100
)

df_updated
```

```python
f = df_updated["Price Increase (%)"] <= 0
df_updated[f]
```

```python
plt.plot(df_updated["Year"], df_updated["Min price"], label="Min price")
plt.plot(df_updated["Year"], df_updated["Max price"], label="Max price")
plt.legend()
plt.grid(True)
plt.show()
```

```python
df_updated.plot(
    x="Year",
    y=["Min price", "Max price"],
    grid=True,
    ylabel="Price (EUR)",
    title="Oktoberfest beer price",
)
plt.show()
```

```python
df_updated["Min sales"] = (
    df_updated["Beer consumption (million liters)"] * df_updated["Min price"]
)

df_updated["Max sales"] = (
    df_updated["Beer consumption (million liters)"] * df_updated["Max price"]
)
```

```python
df_updated.plot(x="Year", y=["Min sales", "Max sales"], kind="bar")
plt.title("Revenue")
plt.ylabel("Million EUR")
plt.show()
```

```python
x = df_updated["Year"]
width = 0.4

plt.figure(figsize=(11, 5))
plt.bar(x - width / 2, df_updated["Min sales"], width=width, label="Min sales")
plt.bar(x + width / 2, df_updated["Max sales"], width=width, label="Max sales")
plt.title("Revenue")
plt.xlabel("Year")
plt.ylabel("Million EUR")
# plt.xticks(range(1980, 2026, 5))  # Show years every 5 years
# otherwise it is automatically set by matplotlib
plt.legend()
plt.show()
```

```python
cols = ["Year", "VPI", "vpi_annotation", "inflation", "inflation_annotation"]

vpi = pd.read_csv(
    "vpi.csv",
    sep=";",
    decimal=",",
    skiprows=6,
    skipfooter=3,
    names=cols,
    usecols=["Year", "VPI", "inflation"]
)
vpi.head()
```

```python
df_oktoberfest = pd.merge(df_updated, vpi, on="Year", how="outer", indicator=True)
df_oktoberfest.head()
```

```python
base_vpi = df_oktoberfest.loc[df_oktoberfest["Year"] == 2010, "VPI"].iloc[0]
# iloc takes the first argument of the series, which is a scalar value
print(base_vpi)
```

```python
df_oktoberfest["Min price (2010)"] = (
df_oktoberfest["Min price"] * base_vpi / df_oktoberfest["VPI"]
)
df_oktoberfest["Max price (2010)"] = (
df_oktoberfest["Max price"] * base_vpi / df_oktoberfest["VPI"]
)
```

```python
df_oktoberfest.tail(10)
```

```python
df_oktoberfest.plot(
    x="Year",
    y=["Min price", "Max price", "Min price (2010)", "Max price (2010)"],
    color=["red", "orange", "blue", "green"],
    style=["--", "--", "-", "-"], # dashed for nominal, solid for real
    figsize=(10, 6),
    grid=True,
    ylabel="Price (EUR)",
    title="Oktoberfest beer price per Maß: nominal vs real",
)
plt.show()
```
